```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("darkgrid")

%matplotlib inline
```

```python
df=pd.read_excel("supermarket_sales.xlsx")
df.head()
```

```python
df.shape
```

```python
df.describe()
```

```python
plt.figure(figsize=(12,7))
corr = df.select_dtypes(include="number").corr()
sns.heatmap(corr, annot=True)
plt.show()
```

```python
df['City'].value_counts()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(x='City',data=df).set_title("Customer Count",fontsize=15,fontdict={"weight": "bold"})
```

```python
total_sales = df.groupby("City").sum(numeric_only=True)
total_sales["Total"].sort_values(ascending=False)
```

```python
plt.figure(figsize=(7,5))
sns.barplot(x="City",y='Total',data=df,order=["Naypyitaw","Yangon","Mandalay"],palette='bright',estimator=sum).set_title("Total Sales",fontsize=15,fontdict={"weight": "bold"})
```

```python
total_quantity = df.groupby("City")["Quantity"].sum()
total_quantity.sort_values(ascending=False)
```
